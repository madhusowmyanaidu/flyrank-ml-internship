## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Research question:**
Which pages should be reviewed first for a possible content refresh or improvement based on observable search-performance and content signals?

**Decision:**
The decision is which pages should receive higher priority for human review.

**Who acts on the decision:**
An SEO or content specialist would review the prioritized pages and decide whether a page should be refreshed, expanded, protected, monitored, consolidated, or left unchanged.

**Action:**
The eventual system could produce a ranked list of pages with supporting signals or reason codes. The specialist could then investigate the highest-priority pages first.

**Cost of a wrong recommendation:**
A false positive could cause the SEO or content team to waste time and resources improving a page that does not need attention. A false negative could cause a page with a genuine opportunity or performance problem to be overlooked. Therefore, the result should be treated as decision support rather than an automatic decision.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

**What the initial data shows**

The starter dataset contains 30,000 pages. The median number of 90-day impressions is 731, while the median average position is 10.8. The median CTR is 0.07. These numbers show that the dataset contains a substantial number of pages with measurable search visibility and performance differences. This makes it worth investigating whether observable page and search signals can help prioritize which pages should be reviewed first for a possible content refresh. These initial statistics do not prove that a refresh will improve performance; they only show that there is enough measurable variation to investigate the question further.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/madhusowmyanaidu/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url, on_bad_lines="warn")

print("Number of rows:", len(df))
print("Average 90-day impressions:", round(df["impressions_90d"].mean(), 2))
print("Average 90-day sessions:", round(df["sessions_90d"].mean(), 2))
print(
    "Percentage of pages with declining trend:",
    round((df["trend_direction"].eq("down").mean() * 100), 2),
    "%"
)
```

## 4. Careful words: what I can and can't claim
